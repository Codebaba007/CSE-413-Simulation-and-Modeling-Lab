```python
import numpy as np

scalar = 3.7

arr = np.array([1.5, 2.8, 3.7, 4.2, 5.6, 6.9])

print("Scalar:", scalar)
print("Array:", arr)
```

```python
print("Trigonometric")

print("sin:", np.sin(arr))
print("cos:", np.cos(arr))
print("tan:", np.tan(arr))

clip_array = np.clip(arr / 10, -1, 1)

print("arcsin:", np.arcsin(clip_array))
print("arccos:", np.arccos(clip_array))
print("arctan:", np.arctan(arr))
```

```python
print("Exponential")

print("exp:", np.exp(arr))
print("exp2:", np.exp2(arr))
```

```python
print("Power")

print("Square:", np.square(arr))
print("Cube:", np.power(arr,3))
print("Square Root:", np.sqrt(arr))
```

```python
print("Rounding")

print("Round:", np.round(arr))
print("Floor:", np.floor(arr))
print("Ceil:", np.ceil(arr))
```

```python
print("Others")

print("Absolute:", np.abs(arr))
print("Remainder:", np.remainder(arr,2))
```

```python
matrix = np.array([
    [2.3, 4.5],
    [6.1, 8.7],
    [1.9, 3.4]
])
#trigono
print(matrix)
print("sin")
print(np.sin(matrix))

print("cos")
print(np.cos(matrix))

print("tan")
print(np.tan(matrix))
#inver_trig
clip_matrix = np.clip(matrix/10,-1,1)

print(np.arcsin(clip_matrix))
print(np.arccos(clip_matrix))
print(np.arctan(matrix))
#expo
print(np.exp(matrix))
print(np.exp2(matrix))
#log
print(np.log(matrix))
print(np.log10(matrix))
print(np.log2(matrix))
#root
print(np.abs(matrix))
print(np.sqrt(matrix))
#reminder
print(np.remainder(matrix,2))
#rounding
print(np.round(matrix))
print(np.floor(matrix))
print(np.ceil(matrix))
```

```python
matrix = np.array([
    [2.1, 4.7, 6.2],
    [1.8, 8.3, 5.9]
])

print(matrix)

#max_min
print(np.max(matrix))
print(np.argmax(matrix))

print(np.min(matrix))
print(np.argmin(matrix))

#length
print(np.linalg.norm(matrix))

#sort
print(np.sort(matrix, axis=1)[:, ::-1])
print(np.sort(matrix.flatten())[::-1])

#sum_product
print(np.sum(matrix))
print(np.prod(matrix))

#statistics
print(np.median(matrix))
print(np.mean(matrix))
print(np.std(matrix))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.heatmap(matrix, annot=True, cmap="viridis")
plt.title("2x3 Matrix Heatmap")
plt.show()
```
