```python


import numpy as np
from scipy.stats import ttest_ind, t
```

```python


data1 = np.array([18.4, 19.1, 17.9, 18.7, 18.3])
data2 = np.array([20.2, 20.5, 20.1, 20.3, 20.4])

print("Data 1:", data1)
print("Data 2:", data2)
```

```python


data1 = np.array([18.4, 19.1, 17.9, 18.7, 18.3])
data2 = np.array([20.2, 20.5, 20.1, 20.3, 20.4])

print("Data 1:", data1)
print("Data 2:", data2)
```

```python
alpha = 0.05

print("Significance level (alpha):", alpha)
```

```python
# Hypotheses

print("H0: The population means are equal (μ1 = μ2)")
print("H1: The population means are different (μ1 ≠ μ2)")
print("This is a two-tailed test.")
```

```python
# Calculate sample means

mean1 = np.mean(data1)
mean2 = np.mean(data2)

print("Sample Mean of Data 1:", mean1)
print("Sample Mean of Data 2:", mean2)
```

```python
# Calculate sample variances
# ddof=1 gives the sample variance

var1 = np.var(data1, ddof=1)
var2 = np.var(data2, ddof=1)

print("Sample Variance of Data 1:", var1)
print("Sample Variance of Data 2:", var2)
```

```python
# Calculate the difference between the sample means

mean_difference = mean1 - mean2

print("Mean Difference (Data 1 - Data 2):", mean_difference)
```

```python
t_stat, p_value = ttest_ind(data1, data2, equal_var=True)

n1 = len(data1)
n2 = len(data2)

df = n1 + n2 - 2

print("t-statistic:", t_stat)
print("Degrees of Freedom:", df)
print("Two-tailed p-value:", p_value)
```

```python
# Calculate pooled variance

pooled_variance = (
    ((n1 - 1) * var1) + ((n2 - 1) * var2)
) / df

# Calculate pooled standard deviation

pooled_std = np.sqrt(pooled_variance)

# Calculate standard error of the difference

standard_error = pooled_std * np.sqrt((1 / n1) + (1 / n2))

print("Pooled Variance:", pooled_variance)
print("Pooled Standard Deviation:", pooled_std)
print("Standard Error:", standard_error)
```

```python
# Calculate the 95% confidence interval

confidence_level = 0.95

# Critical t-value
critical_t = t.ppf(1 - alpha / 2, df)

# Margin of error
margin_of_error = critical_t * standard_error

# Confidence interval
lower_limit = mean_difference - margin_of_error
upper_limit = mean_difference + margin_of_error

print("Critical t-value:", critical_t)
print("Margin of Error:", margin_of_error)
print("95% Confidence Interval:")
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
# Decision based on the p-value

if p_value < alpha:
    decision = "Reject H0"
    interpretation = (
        "There is statistically significant evidence that the "
        "two population means differ."
    )
else:
    decision = "Fail to Reject H0"
    interpretation = (
        "There is not enough statistical evidence to conclude "
        "that the two population means differ."
    )

print("Decision:", decision)
print("Interpretation:", interpretation)

# Final Results

print("========== TWO-SAMPLE t-TEST RESULTS ==========")
print()
print("Sample Mean (Data 1):", mean1)
print("Sample Mean (Data 2):", mean2)
print("Sample Variance (Data 1):", var1)
print("Sample Variance (Data 2):", var2)
print("Mean Difference:", mean_difference)
print()
print("t-statistic:", t_stat)
print("Degrees of Freedom:", df)
print("Two-tailed p-value:", p_value)
print()
print("95% Confidence Interval:")
print("(", lower_limit, ",", upper_limit, ")")
print()
print("Decision:", decision)
print("Interpretation:", interpretation)
print("================================================")
```
